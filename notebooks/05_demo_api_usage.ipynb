```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve().parents[0] if (Path.cwd() / 'src').exists() else Path.cwd().resolve().parents[1]
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))
print('PROJECT_ROOT =', PROJECT_ROOT)
```

```python
import requests
from src.iris_project.preprocessing import list_images

images = list_images(PROJECT_ROOT / 'data')
if images:
    with open(images[0], 'rb') as f:
        response = requests.post('http://127.0.0.1:8000/predict', files={'file': f})
    print(response.json())
else:
    print('No images found')
```
